```python
# Task 1
def top_items(n, *sales):
    ranked = sorted(sales, key = lambda sales : sales[1], reverse = True)
    return [item for item, amount in ranked[:n]]

print(top_items(2, ("pen", 50), ("book", 300), ("bag", 120)))
print(top_items(5, ("pen", 50), ("book", 300)))
print(top_items(3))
```

```python
# Task 2
class Product:
    def __init__(self, name, price, stock):
        self.name = name
        self.price = price
        self.stock = stock

    def restock(self, qty):
        self.stock += qty

    def sell(self, qty):
        if qty <= self.stock:
            self.stock -= qty
            return True
        else:
            return False
            
    def value(self):
        return self.price * self.stock


pen = Product("pen", 50, 10)
print(pen.sell(4), pen.stock)
print(pen.sell(15), pen.stock)
pen.restock(4)
print(pen.value())
```
